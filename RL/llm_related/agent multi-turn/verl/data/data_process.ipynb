```python
import pandas as pd
```

```python
datas = ['读取文件，统计数字列与非数字列的比例。文件路径如下：test.xlsx']*100
df = pd.DataFrame(datas, columns=['question'])
df.head()
```

```python
train_df = pd.DataFrame(columns=['data_source', 'prompt', 'ability', 'reward_model', 'extra_info', 'raw_prompt', 'agent_name'])
```

```python
data_source = []
prompt = []
ability = []
reward_model = []
extra_info = []
raw_prompt = []
agent_name = []
```

```python
for idx, ques in enumerate(df['question']):
    data_source.append('code')
    p = [{"role": "user", "content": ques}]
    prompt.append(p)
    raw_prompt.append(p)
    ability.append('text')
    reward_model.append({
                "style": "rule",
                "ground_truth": "",
            })
    
    extra_info.append({
                'split': 'train',
                'index': idx,
            })
    agent_name.append('tool_agent')
    
```

```python
train_df['data_source'] = data_source
train_df['prompt'] = prompt
train_df['ability'] = ability
train_df['reward_model'] = reward_model
train_df['extra_info'] = extra_info
train_df['raw_prompt'] = raw_prompt
train_df['agent_name'] = agent_name
```

```python
train_df.to_parquet('./train_code.parquet')
```
